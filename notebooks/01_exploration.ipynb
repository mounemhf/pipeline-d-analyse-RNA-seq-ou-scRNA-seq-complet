# 01 — Exploration of the PBMC 3k analysis outputs

Interactive exploration of the pipeline results. This notebook is **not**
part of the reproducible workflow: all analysis logic lives in
`workflow/scripts/` and is executed by Snakemake.

Prerequisite: run the pipeline first (`snakemake --use-conda --cores 4`)
and start Jupyter from an environment with scanpy installed (e.g.
`workflow/envs/preprocess.yml` plus `jupyterlab`).

```python
from pathlib import Path

import scanpy as sc

sc.settings.set_figure_params(dpi=100, frameon=False)

RESULTS = Path("../results")
adata = sc.read_h5ad(RESULTS / "processed" / "pbmc3k_annotated.h5ad")
adata
```

## Clusters and cell types

Side-by-side UMAPs of the Leiden clusters and the CellTypist annotations
are a quick way to spot clusters whose automated label looks inconsistent.

```python
sc.pl.umap(adata, color=["leiden", "cell_type"], wspace=0.5)
```

## Marker genes

Top markers per cluster (computed by `workflow/scripts/markers.py`).

```python
sc.pl.rank_genes_groups_dotplot(adata, n_genes=5, standard_scale="var")
```
